```python
import os
import glob
import pandas as pd
import concurrent.futures

BASE_PATH = "./FM01bullet/year=2025/month=8/"   # 你的 8 月 parquet 路径

# 递归查找所有 parquet 文件
parquet_files = glob.glob(os.path.join(BASE_PATH, "day=*/**/*.parquet"), recursive=True)
print(f"找到 {len(parquet_files)} 个 parquet 文件")

def load_parquet(file_path):
    """读取单个 parquet 文件"""
    try:
        return pd.read_parquet(file_path)
    except Exception as e:
        print(f"❌ 读取失败 {file_path}: {e}")
        return pd.DataFrame()

# 多线程读取
dfs = []
with concurrent.futures.ThreadPoolExecutor(max_workers=8) as executor:  # 8 个线程，可调
    futures = {executor.submit(load_parquet, f): f for f in parquet_files}
    for future in concurrent.futures.as_completed(futures):
        df = future.result()
        if not df.empty:
            dfs.append(df)

# 合并成一个 DataFrame
final_df = pd.concat(dfs, ignore_index=True)

print(f"合并完成，DataFrame 形状: {final_df.shape}")
```
