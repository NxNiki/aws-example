```python

import pandas as pd
from bituslabs_ds.eda import plot_df_distribution
from bituslabs_ds.athena_utils import execute_query
from bituslabs_ds.config import setup_logging

setup_logging("./.log")

def describe_columns(col_names):

    for col_name in col_names:
        query = f"""
            SELECT {col_name}
            FROM ag_share_data.all_hunter_logs_clean
            WHERE {col_name} IS NOT NULL
            LIMIT 10000
        """

        print(col_name)

        try:
            df = execute_query(query, "ag_share_data")
        except:
            print("Something went wrong")
            continue

        print(df.describe(include='all').T)
        unique_values = df[col_name].unique().tolist()
        print(unique_values[:min(10, len(unique_values))])

        df[col_name] = pd.to_numeric(df[col_name], errors="coerce")
        plot_df_distribution(df, log=True)
```
