```python
import pandas as pd
data = pd.read_csv('./output/wucaishen_grouped_stat_output_24.csv')
```

```python
data.head()
```

```python
data = data[[
    'start_time',
    'bet_median',
    'basepoint_median',
    'payout_median',
    'profit_median',
    'delta_t_median',
    'delta_bet_median',
    'delta_profit_median',
    'streak_median',
    'win_streak_median',
    'lose_streak_median',
    'rtp_mean']]
```

```python

from bituslabs_ds.eda import plot_df_distribution
plot_df_distribution(data.iloc[:, 1:], log=True)
```

```python

from bituslabs_ds.eda import plot_seasonality

plot_seasonality(data['start_time'], data[['basepoint_median']])
plot_seasonality(data['start_time'], data[['bet_median', 'payout_median', 'profit_median']])
plot_seasonality(data['start_time'], data[['delta_t_median']])
plot_seasonality(data['start_time'], data[['delta_bet_median', 'delta_profit_median']])
plot_seasonality(data['start_time'], data[['win_streak_median', 'lose_streak_median']])
plot_seasonality(data['start_time'], data[['streak_median']])
plot_seasonality(data['start_time'], data[['rtp_mean']])
```
