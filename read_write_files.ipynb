```python

import boto3
import pandas as pd
import io

# Define your bucket and object key
bucket_name = 'xin-config'
object_key = 'example.csv'  

# Create an S3 client
s3 = boto3.client('s3')

# Fetch the object from S3
response = s3.get_object(Bucket=bucket_name, Key=object_key)

# Read the body of the response into a pandas DataFrame
df = pd.read_csv(io.BytesIO(response['Body'].read()))

# Display the first few rows
print(df.head())
```

# write data to sagemaker default s3 bucket:

```python

import boto3
import sagemaker

session = sagemaker.Session()
bucket = session.default_bucket()

df.to_csv('data.csv', index=False)
s3_client = boto3.client('s3')
s3_client.upload_file('data.csv', bucket, 'example-data/data.csv')

s3_path = f"s3://{bucket}/example-data/"
print("S3 path:", s3_path)
```
